```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())
```

```python
# Missing values
print("Missing Values:")
print(df.isnull().sum().sum())

# Duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# Unique values in important categorical columns
print("\nAttrition:")
print(df["Attrition"].value_counts())

print("\nDepartment:")
print(df["Department"].value_counts())

print("\nJob Role:")
print(df["JobRole"].value_counts())

print("\nOvertime:")
print(df["OverTime"].value_counts())
```

```python
# Check how many unique values each column contains
unique_counts = df.nunique().sort_values()

print(unique_counts)
```

```python
# Check columns with only one unique value
constant_columns = [col for col in df.columns if df[col].nunique() == 1]

print("Constant columns:")
print(constant_columns)
```

```python
# Check ID-like columns
print("EmployeeNumber unique values:", df["EmployeeNumber"].nunique())
print("EmployeeCount unique values:", df["EmployeeCount"].nunique())
print("StandardHours unique values:", df["StandardHours"].nunique())
print("Over18 unique values:", df["Over18"].nunique())
```

```python
# Create a cleaned copy of the dataset
df_clean = df.copy()

# Remove identifier and constant columns that are not useful for analysis
columns_to_remove = [
    "EmployeeNumber",
    "EmployeeCount",
    "StandardHours",
    "Over18"
]

df_clean = df_clean.drop(columns=columns_to_remove)

print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)

print("\nRemaining columns:")
print(df_clean.columns.tolist())
```

```python
# Overall workforce KPIs

total_employees = len(df_clean)
employees_left = (df_clean["Attrition"] == "Yes").sum()
employees_stayed = (df_clean["Attrition"] == "No").sum()
attrition_rate = (employees_left / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print("Employees Who Stayed:", employees_stayed)
print("Attrition Rate: {:.2f}%".format(attrition_rate))
```

```python
# Department-wise attrition analysis

department_attrition = pd.crosstab(
    df_clean["Department"],
    df_clean["Attrition"]
)

department_attrition["Attrition Rate (%)"] = (
    department_attrition["Yes"] /
    department_attrition.sum(axis=1) * 100
).round(2)

display(department_attrition)
```

```python
# Job role-wise attrition analysis

jobrole_attrition = pd.crosstab(
    df_clean["JobRole"],
    df_clean["Attrition"]
)

jobrole_attrition["Attrition Rate (%)"] = (
    jobrole_attrition["Yes"] /
    jobrole_attrition.sum(axis=1) * 100
).round(2)

display(jobrole_attrition.sort_values("Attrition Rate (%)", ascending=False))
```

```python
# Overtime vs attrition

overtime_attrition = pd.crosstab(
    df_clean["OverTime"],
    df_clean["Attrition"]
)

overtime_attrition["Attrition Rate (%)"] = (
    overtime_attrition["Yes"] /
    overtime_attrition.sum(axis=1) * 100
).round(2)

display(overtime_attrition)
```

```python
# Attrition by Department

department_rate = (
    df_clean.groupby("Department")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
department_rate.plot(kind="bar")
plt.title("Employee Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Job Satisfaction vs Attrition

satisfaction_attrition = pd.crosstab(
    df_clean["JobSatisfaction"],
    df_clean["Attrition"]
)

satisfaction_attrition["Attrition Rate (%)"] = (
    satisfaction_attrition["Yes"] /
    satisfaction_attrition.sum(axis=1) * 100
).round(2)

display(satisfaction_attrition)
```

```python
# Monthly Income vs Attrition

income_attrition = (
    df_clean.groupby("Attrition")["MonthlyIncome"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(2)
)

display(income_attrition)
```

```python
# Years at Company vs Attrition

tenure_attrition = pd.crosstab(
    df_clean["YearsAtCompany"],
    df_clean["Attrition"]
)

tenure_attrition["Attrition Rate (%)"] = (
    tenure_attrition["Yes"] /
    tenure_attrition.sum(axis=1) * 100
).round(2)

display(tenure_attrition.head(15))
```

```python
# Create tenure groups for analysis and Power BI

bins = [-1, 2, 5, 10, 20, float("inf")]
labels = ["0-2 Years", "3-5 Years", "6-10 Years", "11-20 Years", "20+ Years"]

df_clean["TenureGroup"] = pd.cut(
    df_clean["YearsAtCompany"],
    bins=bins,
    labels=labels
)

tenure_group_attrition = pd.crosstab(
    df_clean["TenureGroup"],
    df_clean["Attrition"]
)

tenure_group_attrition["Attrition Rate (%)"] = (
    tenure_group_attrition["Yes"] /
    tenure_group_attrition.sum(axis=1) * 100
).round(2)

display(tenure_group_attrition)
```

```python
# Age group analysis

age_bins = [17, 25, 35, 45, 55, 100]
age_labels = ["18-25", "26-35", "36-45", "46-55", "56+"]

df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=age_bins,
    labels=age_labels
)

age_attrition = pd.crosstab(
    df_clean["AgeGroup"],
    df_clean["Attrition"]
)

age_attrition["Attrition Rate (%)"] = (
    age_attrition["Yes"] /
    age_attrition.sum(axis=1) * 100
).round(2)

display(age_attrition)
```

```python
# Save the cleaned dataset for SQL and Power BI

df_clean.to_csv("employee_attrition_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
print("Shape:", df_clean.shape)
```

```python
import sqlite3

# Create an in-memory SQL database
conn = sqlite3.connect(":memory:")

# Load cleaned dataset into SQL
df_clean.to_sql(
    "employee_attrition",
    conn,
    index=False,
    if_exists="replace"
)

print("Dataset loaded into SQL successfully.")
```

```python
query = """
SELECT
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Employees_Left,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Attrition_Rate
FROM employee_attrition;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Employees_Left,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Attrition_Rate
FROM employee_attrition
GROUP BY Department
ORDER BY Attrition_Rate DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    OverTime,
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Employees_Left,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Attrition_Rate
FROM employee_attrition
GROUP BY OverTime
ORDER BY Attrition_Rate DESC;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    JobSatisfaction,
    COUNT(*) AS Total_Employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS Employees_Left,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS Attrition_Rate
FROM employee_attrition
GROUP BY JobSatisfaction
ORDER BY JobSatisfaction;
"""

display(pd.read_sql_query(query, conn))
```

```python
query = """
SELECT
    Attrition,
    COUNT(*) AS Employee_Count,
    ROUND(AVG(MonthlyIncome), 2) AS Average_Income,
    ROUND(MIN(MonthlyIncome), 2) AS Minimum_Income,
    ROUND(MAX(MonthlyIncome), 2) AS Maximum_Income
FROM employee_attrition
GROUP BY Attrition;
"""

display(pd.read_sql_query(query, conn))
```
